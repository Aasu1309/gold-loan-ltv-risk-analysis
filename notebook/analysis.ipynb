```python
#After removing duplicates and converting purity to a karat number (916 = 22, 833 = 20, 750 = 18, 999 = 24),
# return karat and loans for each karat

karat={'916':22,'833':20,'750':18,'999':24}

def to_karat(p):
    num=''.join(i for i in str(p) if i.isdigit())
    return karat.get(num,int(num))

gold_loans['purity']=gold_loans['purity'].apply(to_karat)
```

```python
import pandas as pd
import numpy as np

# Load dataset
loans = pd.read_csv('/data/gold_loans.csv')

# Remove duplicate records
loans = loans.drop_duplicates()

# Extract numeric values from purity
loans['purity_num'] = (
    loans['purity']
    .astype(str)
    .str.extract(r'(\d+)')[0]
    .astype(float)
)

# Convert purity into karat
mapping = {
    916: 22,
    833: 20,
    750: 18,
    999: 24,
    22: 22,
    20: 20,
    18: 18,
    24: 24
}

loans['karat'] = loans['purity_num'].map(mapping)

# Count loans by karat
answer = (
    loans.groupby('karat')
    .size()
    .reset_index(name='loans')
)

print(answer)
```

```python
# Remove Duplicates records

loans = loans.drop_duplicates()

# Count missing gold weight

answer = loans['gold_weight_gm'].isnull().sum()

print(answer)
```

```python
# Remove Duplicates records

loans = loans.drop_duplicates()

# Count loans where principal os 0 or less

answer = (loans['principal']<= 0).sum()

print(answer)
```

```python
# Remove Duplicates 

loans = loans.drop_duplicates()

# Count active loans (case-insentive)

answer = (loans['status'].astype(str).str.strip().str.lower().eq('active').sum()
)

print(answer)
```

```python
# Remove Duplicates

loans = loans.drop_duplicates()

# Filter active loans

active_loans = loans[
    loans['status'].astype(str).str.strip().str.lower() == 'active'
]

# Count loans with missing gold weight OR invalid principal

answer = (
    active_loans['gold_weight_gm'].isna() |

    (active_loans['principal']<= 0)).sum()

print(answer)
```

```python
# Convert date columns

gold_rates['rate_date'] = pd.to_datetime(gold_rates['rate_date'])

# Filter data for 2026

filtered = gold_rates[
    gold_rates['rate_date'].dt.year == 2026
]

# Find date of highest 24K gold rate

peak_index = filtered['rate_24k_per_gram'].idxmax()

peak_date = filtered.loc[peak_index, 'rate_date']

# Create final answer

answer = pd.DataFrame({'peak_date': [peak_date.strftime('%Y-%m-%d')]
})

print(answer)
```

```python
# convert rate_date into date_time format

gold_rates['rate_date'] = pd.to_datetime(gold_rates['rate_date'])

# Assessment date

assessment_date = pd.timestamp('2026-10'15')

# Filter rates up to assessment date

filtered = gold_rates[gold_rates['rate_date'] <= assessment_date]

# Find the latest available rate 

filtered = filtered.sort_values('rate_date')

answer = filtered.iloc[-1] ['rate_24k_per_gram']

print(answer)
```
