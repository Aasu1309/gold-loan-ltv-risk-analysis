```python
# Q.11 Loans by karat

#After removing duplicates and converting purity to a karat number (916 = 22, 833 = 20, 750 = 18, 999 = 24),
# return karat and loans for each karat

karat={'916':22,'833':20,'750':18,'999':24}

def to_karat(p):
    num=''.join(i for i in str(p) if i.isdigit())
    return karat.get(num,int(num))

gold_loans['purity']=gold_loans['purity'].apply(to_karat)
```

```python
# Q.12 Missing gold weight

# After removing duplicates, how many loans have no gold weight recorded?

import pandas as pd
import numpy as np

# Load dataset
loans = pd.read_csv('/data/gold_loans.csv')

# Remove duplicate records
loans = loans.drop_duplicates()

# Extract numeric values from purity
loans['purity_num'] = (
    loans['purity']
    .astype(str)
    .str.extract(r'(\d+)')[0]
    .astype(float)
)

# Convert purity into karat
mapping = {
    916: 22,
    833: 20,
    750: 18,
    999: 24,
    22: 22,
    20: 20,
    18: 18,
    24: 24
}

loans['karat'] = loans['purity_num'].map(mapping)

# Count loans by karat
answer = (
    loans.groupby('karat')
    .size()
    .reset_index(name='loans')
)

print(answer)
```

```python
# Q.14 Active loans

# How many active loans are there after removing duplicates (status 'active' in any letter case)?

# Remove Duplicates records

loans = loans.drop_duplicates()

# Count loans where principal os 0 or less

answer = (loans['principal']<= 0).sum()

print(answer)
```

```python
# Q.16 Gold rate history

# How many daily rates are in gold_rates?

# Remove Duplicates

loans = loans.drop_duplicates()

# Filter active loans

active_loans = loans[
    loans['status'].astype(str).str.strip().str.lower() == 'active'
]

# Count loans with missing gold weight OR invalid principal

answer = (
    active_loans['gold_weight_gm'].isna() |

    (active_loans['principal']<= 0)).sum()

print(answer)
```

```python
# Q.18 Date of the peak

# Return one row with peak_date: the date of the highest 24K rate in 2026.

# Convert date columns

gold_rates['rate_date'] = pd.to_datetime(gold_rates['rate_date'])

# Filter data for 2026

filtered = gold_rates[
    gold_rates['rate_date'].dt.year == 2026
]

# Find date of highest 24K gold rate

peak_index = filtered['rate_24k_per_gram'].idxmax()

peak_date = filtered.loc[peak_index, 'rate_date']

# Create final answer

answer = pd.DataFrame({'peak_date': [peak_date.strftime('%Y-%m-%d')]
})

print(answer)
```

```python
# Q.19 Rate on the assessment date
# Which 24K rate (₹ per gram) applies on 15 October 2026?

# Convert date column into datetime
gold_rates['rate_date'] = pd.to_datetime(gold_rates['rate_date'])

# Assessment date
assessment_date = pd.Timestamp('2026-10-15')

# Filter rates up to assessment date
filtered = gold_rates[
    gold_rates['rate_date'] <= assessment_date
]

# Sort by date
filtered = filtered.sort_values('rate_date')

# Find latest available 24K gold rate
answer = filtered.iloc[-1]['rate_24k_per_gram']

print(answer)

```

```python
# Q.20 Fall from the peak

# By what percentage has the 24K rate fallen from its 2026 peak to the assessment date? Round to 2 decimals.

# Convert date columns into datetime
gold_rates['rate_date'] = pd.to_datetime(gold_rates['rate_date'])

# Find 2026 gold rates
gold_2026 = gold_rates[gold_rates['rate_date'].dt.year == 2026]

# Find highest 24k gold rate
peak_rate = gold_2026['rate_24k_per_gram'].max()

# Assessment date
assessment_date = pd.Timestamp('2026-10-15')

# Find latest available rate on or before assessment date 
assessment_rates = gold_rates[gold_rates['rate_date'] <= assessment_date].sort_values('rate_date')

assessment_rate = assessment_rates.iloc[-1]['rate_24k_per_gram']

# Calculate percentage fall
answer = ((peak_rate - assessment_rate) / peak_rate) * 100

# Round to 2 decimal places
answer = round(answer, 2)

print(answer)

```

```python
# Q.23 Total disbursed

# After removing duplicates, what is the total principal disbursed across all loans with a principal above 0 (₹)?

# Remove duplicate loan records
cleaned = gold_loans.drop_duplicates(subset='loan_id')

# Filter principal greater than 0
filtered = cleaned[cleaned['principal'] > 0]

# Calculated total principal disbursed
answer = filtered['principal'].sum()

print(answer)
```

```python
# Q.24 Average active loan

# What is the average principal (₹, rounded to 0 decimals) of assessable active loans?

# Remove duplicate loan records
cleaned = gold_loans.drop_duplicates(subset='loan_id')

# Filter active loans with positive principal
filtered = cleaned[
    (cleaned['status'].str.lower() == 'active') &
    (cleaned['principal'] > 0)
]

# Calculate average principal
answer = round(filtered['principal'].mean())

print(answer)
```

```python
# Q.27 Loans disbursed in 2026

# After removing duplicates, how many loans were disbursed in 2026?

# Remove duplicate loan records
cleaned = gold_loans.drop_duplicates(
    subset='loan_id'
).copy()

# Convert date column into datetime format
cleaned['disbursal_date'] = pd.to_datetime(
    cleaned['disbursal_date'],
    errors='coerce'
)

# Filter loans disbursed in 2026
filtered = cleaned[
    cleaned['disbursal_date'].dt.year == 2026
]

# Count total loans
answer = len(filtered)

print(answer)
```

```python
# Q.28 Interest rate by scheme

# After removing duplicates, return scheme and avg_rate (average interest_rate_pct, rounded to 2 decimals).

# Expected columns: scheme, avg_rate · any row order

# Remove duplicate loan records
cleaned = gold_loans.drop_duplicates(subset='loan_id')

# Calculate average interest rate by scheme
answer = (
    cleaned.groupby('scheme', as_index=False)
    .agg(avg_rate=('interest_rate_pct', 'mean'))
)

# Round average rate to 2 decimal places
answer['avg_rate'] = answer['avg_rate'].round(2)

# Display result
print(answer)
```
