```python
# Q.11 Loans by karat

#After removing duplicates and converting purity to a karat number (916 = 22, 833 = 20, 750 = 18, 999 = 24),
# return karat and loans for each karat

karat={'916':22,'833':20,'750':18,'999':24}

def to_karat(p):
    num=''.join(i for i in str(p) if i.isdigit())
    return karat.get(num,int(num))

gold_loans['purity']=gold_loans['purity'].apply(to_karat)
```

```python
# Q.12 Missing gold weight

# After removing duplicates, how many loans have no gold weight recorded?

import pandas as pd
import numpy as np

# Load dataset
loans = pd.read_csv('/data/gold_loans.csv')

# Remove duplicate records
loans = loans.drop_duplicates()

# Extract numeric values from purity
loans['purity_num'] = (
    loans['purity']
    .astype(str)
    .str.extract(r'(\d+)')[0]
    .astype(float)
)

# Convert purity into karat
mapping = {
    916: 22,
    833: 20,
    750: 18,
    999: 24,
    22: 22,
    20: 20,
    18: 18,
    24: 24
}

loans['karat'] = loans['purity_num'].map(mapping)

# Count loans by karat
answer = (
    loans.groupby('karat')
    .size()
    .reset_index(name='loans')
)

print(answer)
```

```python
# Q.19 Rate on the assessment date
# Which 24K rate (₹ per gram) applies on 15 October 2026?

# Convert date column into datetime
gold_rates['rate_date'] = pd.to_datetime(gold_rates['rate_date'])

# Assessment date
assessment_date = pd.Timestamp('2026-10-15')

# Filter rates up to assessment date
filtered = gold_rates[
    gold_rates['rate_date'] <= assessment_date
]

# Sort by date
filtered = filtered.sort_values('rate_date')

# Find latest available 24K gold rate
answer = filtered.iloc[-1]['rate_24k_per_gram']

print(answer)

```

```python
# Q.20 Fall from the peak

# By what percentage has the 24K rate fallen from its 2026 peak to the assessment date? Round to 2 decimals.

# Convert date columns into datetime
gold_rates['rate_date'] = pd.to_datetime(gold_rates['rate_date'])

# Find 2026 gold rates
gold_2026 = gold_rates[gold_rates['rate_date'].dt.year == 2026]

# Find highest 24k gold rate
peak_rate = gold_2026['rate_24k_per_gram'].max()

# Assessment date
assessment_date = pd.Timestamp('2026-10-15')

# Find latest available rate on or before assessment date 
assessment_rates = gold_rates[gold_rates['rate_date'] <= assessment_date].sort_values('rate_date')

assessment_rate = assessment_rates.iloc[-1]['rate_24k_per_gram']

# Calculate percentage fall
answer = ((peak_rate - assessment_rate) / peak_rate) * 100

# Round to 2 decimal places
answer = round(answer, 2)

print(answer)

```

```python
# Q.49 Worst branch in each region

# For each region, return the branch with the highest breach rate (% of its assessable active loans in breach): region, branch_name, breach_rate_pct (rounded to 2).

#Expected columns: region, branch_name, breach_rate_pct · any row order
branch_breach=book.merge(branch,on='branch_id').groupby(['region','branch_name']).agg(breaching=('breach_new','sum'),loans_new=('loan_id','size')).reset_index()
branch_breach['breach_rate_pct']=round(branch_breach['breaching']/branch_breach['loans_new']*100,2)
answer=branch_breach[['region','branch_name','breach_rate_pct']]
answer=answer.sort_values(['region','breach_rate_pct','branch_name'],ascending=[True,False,True]).groupby('region').head(1)
```
